```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd

RESULTS_DIR = Path('results')
GRAPHS_DIR = Path('graphs')
GRAPHS_DIR.mkdir(exist_ok=True)
```

```python
df = pd.read_csv(RESULTS_DIR / 'ejercicio3_diseccion.csv')
n_proc = df['n_procesos'].to_numpy()
bw_medido = df['bw_MBs'].to_numpy()
df
```

```python
b_base = bw_medido[0]
bw_teorico = (n_proc / 2.0) * b_base

plt.figure(figsize=(7, 4.5))
plt.plot(n_proc, bw_medido, 'o-', color='crimson', label='Ancho de banda medido')
plt.plot(n_proc, bw_teorico, 's--', color='navy', label='Ancho de banda teórico ((n/2) · b)')
plt.xticks(n_proc)
plt.xlabel('Número de procesadores (n)')
plt.ylabel('Ancho de banda de disección (MB/s)')
plt.title('Ancho de Banda de Disección vs. Número de Procesadores')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.savefig(GRAPHS_DIR / 'e3_ancho_de_banda_numero_procesadores.png', dpi=300, bbox_inches='tight')
plt.show()
```
